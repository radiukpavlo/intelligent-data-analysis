**Ви виконаєте такі завдання:**

1. **Контентні рекомендації (content-based filtering):** побудуєте моделі, які пропонують об’єкти на основі їхніх характеристик – жанрів фільмів, тегів виконавців, описів товарів, – і перевірите, наскільки ці пропозиції релевантні, за часткою спільних ознак із запитом і внутрішньосписочною схожістю.
2. **Колаборативна фільтрація (collaborative filtering):** побудуєте моделі на основі пам’яті (memory-based), що шукають схожих користувачів (user-based) або схожі об’єкти (item-based) за вподобаннями користувачів – оцінками, прослуховуваннями, покупками [1, 9], – і оціните їх за `RMSE`, `MAE`, `precision@k`, `recall@k` та іншими метриками.
3. **Факторизація матриць (matrix factorization):** побудуєте моделі, які розкладають матрицю взаємодій на добуток матриць латентних факторів (SVD, SVD зі зміщеннями, NMF, ALS), щоб спрогнозувати вподобання користувачів [2, 3], і оціните їх за `RMSE`, `MAE`, `precision@k`, `recall@k`, `MAP@k` і `NDCG@k`.
4. **Ранжування та оцінювання:** порівняєте способи ранжування рекомендацій – поєднання популярності з прогнозами моделей, переранжування для різноманітності, калібрування – і перевірите їх за `precision@k`, `recall@k`, `MAP@k` і `NDCG@k` [5–8].
5. **Гібридні та ансамблеві системи:** поєднаєте контентну та колаборативну фільтрацію або скомбінуєте кілька моделей (зважений, перемикальний, змішаний і каскадний гібриди, комбінування ознак, стекінг, голосування) [4] і перевірите, чи працюють вони краще за окремі моделі.

**Критерії оцінювання:**

Формули нижче обов’язкові для всіх варіантів; готові функції наведено в [розділі 11.0](#lab-11.0).

- **Метрики помилки прогнозу оцінок** – лише для явних оцінок, на множині $T$ відкладених пар користувач–об’єкт: $\mathrm{RMSE}=\sqrt{\frac{1}{|T|}\sum_{(u,i)\in T}(r_{ui}-\hat r_{ui})^2}$, $\mathrm{MAE}=\frac{1}{|T|}\sum_{(u,i)\in T}|r_{ui}-\hat r_{ui}|$.
- **Метрики ранжування** – для списку $L_u$ з перших $k$ рекомендацій користувача $u$ і множини $R_u$ його релевантних відкладених об’єктів; значення усереднюйте за користувачами, для яких $R_u$ непорожня:
    - $\mathrm{precision@}k=\frac{|L_u\cap R_u|}{k}$, $\mathrm{recall@}k=\frac{|L_u\cap R_u|}{|R_u|}$; частка влучань $\mathrm{HR@}k$ (hit rate) дорівнює 1, якщо $|L_u\cap R_u|>0$, і 0 в іншому разі;
    - $\mathrm{AP@}k=\frac{1}{\min(k,|R_u|)}\sum_{j=1}^{k}\mathrm{rel}_j\cdot\mathrm{precision@}j$, де $\mathrm{rel}_j=1$, якщо об’єкт на позиції $j$ релевантний, інакше $\mathrm{rel}_j=0$; $\mathrm{MAP@}k$ – середнє значення $\mathrm{AP@}k$ за користувачами;
    - $\mathrm{DCG@}k=\sum_{j=1}^{k}\frac{2^{g_j}-1}{\log_2(j+1)}$ і $\mathrm{NDCG@}k=\frac{\mathrm{DCG@}k}{\mathrm{IDCG@}k}$, де $g_j$ – релевантність об’єкта на позиції $j$ (для бінарної релевантності $g_j=\mathrm{rel}_j$), а $\mathrm{IDCG@}k$ – значення $\mathrm{DCG@}k$ для ідеального впорядкування відкладених об’єктів користувача.
- **Метрики покриття, популярності й різноманітності** (beyond-accuracy metrics) – для варіантів про популярність і різноманітність: покриття каталогу (coverage) – частка об’єктів каталогу, що хоча б раз потрапили до рекомендацій; середня популярність рекомендацій (ARP, average recommendation popularity) – середня кількість взаємодій рекомендованих об’єктів у навчальній частині; внутрішньосписочна схожість (intra-list similarity) – середня попарна косинусна схожість об’єктів одного списку.
- Скрізь $k=10$ (списки `top-10`); значення для $k=5$ і $k=20$ наводьте, лише якщо цього вимагає варіант.

```python
# Базові бібліотеки курсу (встановіть, якщо їх немає в середовищі):
#   pip install numpy pandas scipy scikit-learn matplotlib seaborn
#
# Додаткові бібліотеки для окремих варіантів (встановлюйте лише потрібні):
#   conda install -c conda-forge scikit-surprise   # модуль surprise: SVD, SVD зі зміщеннями, NMF, KNNWithMeans, BaselineOnly
#   conda install -c conda-forge implicit          # ALS і BPR для неявного зворотного зв’язку
#   pip install pyspark                            # ALS для явних оцінок (pyspark.ml.recommendation.ALS; потрібна Java 17 або новіша)
#   pip install jieba                              # сегментація китайських текстів (варіант 16)
#
# Пакети scikit-surprise та implicit містять скомпільовані модулі; якщо для вашої версії Python готової збірки немає,
# створіть окреме середовище, наприклад:
#   conda create -n recsys -c conda-forge python=3.11 numpy pandas scipy scikit-learn matplotlib seaborn scikit-surprise implicit

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.sparse import csr_matrix
from sklearn.metrics.pairwise import cosine_similarity

sns.set_theme(style="whitegrid")
RANDOM_STATE = 17
```

```python
# Допоміжні функції: розбиття даних, базова модель за популярністю та метрики (формули – у критеріях оцінювання)


def split_by_user(df, user_col="user_id", time_col=None, test_size=0.2, min_interactions=5,
                  random_state=RANDOM_STATE):
    """Поділ взаємодій кожного користувача на навчальну й тестову частини.

    Залишає користувачів щонайменше з min_interactions взаємодіями. Якщо задано time_col, у тестову частину
    потрапляють останні за часом test_size взаємодій кожного користувача (щонайменше одна), інакше – випадкові.
    """
    counts = df.groupby(user_col, observed=True)[user_col].transform("size")
    df = df[counts >= min_interactions]
    if time_col is not None:
        df = df.sort_values([user_col, time_col], kind="mergesort")
    else:
        df = df.sample(frac=1.0, random_state=random_state).sort_values(user_col, kind="mergesort")
    position = df.groupby(user_col, observed=True).cumcount()
    size = df.groupby(user_col, observed=True)[user_col].transform("size")
    n_test = np.maximum(1, np.floor(size * test_size)).astype(int)
    is_test = position >= size - n_test
    return df[~is_test].copy(), df[is_test].copy()


def popularity_recommendations(train, users, user_col="user_id", item_col="item_id", k=10):
    """Базова модель: k найпопулярніших у навчальній частині об’єктів, яких користувач ще не мав."""
    ranking = train[item_col].value_counts().index.tolist()
    seen = train.groupby(user_col, observed=True)[item_col].agg(set).to_dict()
    recommendations = {}
    for user in users:
        user_seen = seen.get(user, set())
        top = []
        for item in ranking:
            if item not in user_seen:
                top.append(item)
                if len(top) == k:
                    break
        recommendations[user] = top
    return recommendations


def precision_at_k(recommended, relevant, k=10):
    """Частка релевантних об’єктів серед перших k рекомендацій (знаменник – завжди k)."""
    relevant = set(relevant)
    return sum(item in relevant for item in list(recommended)[:k]) / k


def recall_at_k(recommended, relevant, k=10):
    """Частка релевантних об’єктів користувача, які потрапили до перших k рекомендацій."""
    relevant = set(relevant)
    if not relevant:
        return np.nan
    return sum(item in relevant for item in list(recommended)[:k]) / len(relevant)


def hit_rate_at_k(recommended, relevant, k=10):
    """1, якщо серед перших k рекомендацій є хоча б один релевантний об’єкт, інакше 0."""
    relevant = set(relevant)
    return float(any(item in relevant for item in list(recommended)[:k]))


def average_precision_at_k(recommended, relevant, k=10):
    """AP@k з нормуванням на min(k, кількість релевантних об’єктів)."""
    relevant = set(relevant)
    if not relevant:
        return np.nan
    hits, score = 0, 0.0
    for pos, item in enumerate(list(recommended)[:k], start=1):
        if item in relevant:
            hits += 1
            score += hits / pos
    return score / min(k, len(relevant))


def ndcg_at_k(recommended, relevant, k=10):
    """NDCG@k; relevant – множина об’єктів (бінарна релевантність) або словник {об’єкт: релевантність g}."""
    gains = relevant if isinstance(relevant, dict) else dict.fromkeys(relevant, 1.0)
    if not gains:
        return np.nan
    dcg = sum((2.0 ** gains.get(item, 0.0) - 1.0) / np.log2(pos + 1)
              for pos, item in enumerate(list(recommended)[:k], start=1))
    ideal = sorted(gains.values(), reverse=True)[:k]
    idcg = sum((2.0 ** g - 1.0) / np.log2(pos + 1) for pos, g in enumerate(ideal, start=1))
    return dcg / idcg if idcg > 0 else np.nan


def evaluate_top_k(recommendations, test_relevant, k=10):
    """Середні ранжувальні метрики за користувачами, які мають хоча б один релевантний відкладений об’єкт.

    recommendations – словник {користувач: список об’єктів у порядку спадання оцінки моделі};
    test_relevant – словник {користувач: множина релевантних об’єктів тестової частини}.
    """
    rows = []
    for user, relevant in test_relevant.items():
        if len(relevant) == 0:
            continue
        recs = list(recommendations.get(user, []))[:k]
        rows.append({
            f"precision@{k}": precision_at_k(recs, relevant, k),
            f"recall@{k}": recall_at_k(recs, relevant, k),
            f"hit_rate@{k}": hit_rate_at_k(recs, relevant, k),
            f"map@{k}": average_precision_at_k(recs, relevant, k),
            f"ndcg@{k}": ndcg_at_k(recs, relevant, k),
        })
    return pd.DataFrame(rows).mean()


def catalog_coverage(recommendations, n_items):
    """Покриття каталогу: частка об’єктів, які хоча б раз потрапили до рекомендацій."""
    recommended = {item for recs in recommendations.values() for item in recs}
    return len(recommended) / n_items


def average_recommendation_popularity(recommendations, item_popularity):
    """ARP: середня за користувачами популярність рекомендованих об’єктів (кількість взаємодій у навчальній частині)."""
    per_user = [np.mean([item_popularity.get(item, 0) for item in recs])
                for recs in recommendations.values() if len(recs) > 0]
    return float(np.mean(per_user)) if per_user else np.nan


def intra_list_similarity(item_rows, features):
    """Середня попарна косинусна схожість об’єктів списку; item_rows – номери рядків матриці ознак features."""
    item_rows = list(item_rows)
    n = len(item_rows)
    if n < 2:
        return np.nan
    sim = cosine_similarity(features[item_rows])
    return float((sim.sum() - np.trace(sim)) / (n * (n - 1)))


def jaccard(a, b):
    """Коефіцієнт Жаккара двох множин ознак (жанрів, тегів, категорій)."""
    a, b = set(a), set(b)
    return len(a & b) / len(a | b) if a | b else np.nan
```

```python
# Приклад: завантаження MovieLens ml-latest-small і базова модель за популярністю за протоколом Завдань 2–5
import zipfile
from pathlib import Path
from urllib.request import urlretrieve

ML_URL = "https://files.grouplens.org/datasets/movielens/ml-latest-small.zip"
data_dir = Path("ml-latest-small")  # локальна копія поруч із ноутбуком, якщо є
if not (data_dir / "ratings.csv").exists():
    archive, _ = urlretrieve(ML_URL, "ml-latest-small.zip")
    with zipfile.ZipFile(archive) as zf:
        zf.extractall(".")

ratings = pd.read_csv(data_dir / "ratings.csv")
movies = pd.read_csv(data_dir / "movies.csv")

n_users, n_items = ratings["userId"].nunique(), ratings["movieId"].nunique()
print(f"Оцінок: {len(ratings)}, користувачів: {n_users}, фільмів з оцінками: {n_items}, "
      f"розрідженість: {1 - len(ratings) / (n_users * n_items):.4f}")

# Часове розбиття для кожного користувача; релевантні – відкладені оцінки ≥ 4
train, test = split_by_user(ratings, user_col="userId", time_col="timestamp")
test_relevant = test[test["rating"] >= 4].groupby("userId")["movieId"].agg(set).to_dict()
popular = popularity_recommendations(train, test_relevant.keys(), user_col="userId", item_col="movieId", k=10)

print(evaluate_top_k(popular, test_relevant, k=10).round(4))
print("Покриття каталогу:", round(catalog_coverage(popular, n_items), 4))
print("ARP:", round(average_recommendation_popularity(popular, train["movieId"].value_counts()), 1))
```

- Оцінювання: ранжувальні метрики, названі у варіанті (формули – у критеріях оцінювання); `RMSE` – лише тоді, коли стратегія прогнозує оцінку, і лише на відкладених парах; для варіантів про популярність, різноманітність і калібрування – також покриття каталогу, ARP, внутрішньосписочна схожість або міра, визначена у варіанті. Базова модель – популярність.
- Відтворюваність: параметри стратегій (ваги, параметр згасання, $\lambda$, $C$ тощо) добирайте на валідаційній частині; випадкові операції виконуйте з `random_state=17`.
- Звіт: порівняльна таблиця метрик усіх стратегій і базової моделі, щонайменше одна візуалізація (наприклад, стовпчаста діаграма метрик або залежність метрики від параметра стратегії), два списки `top-10` для одного користувача з позначенням спільних об’єктів, висновок (3–5 речень) з відповіддю на запитання варіанта.
